# How to choose a machine learning model and evaluation metrics

# ----------------Selecting the appropriate machine learning model--------------------

The choice of model depends on the structure of the data and the task at hand:

# Step1.Identify target variable type(y)

**Supervised Regression:** if y is continous numerical variables(e.g fuel efficiency in miles per gallon, house price, stock value), select regression models such as LinearRegression, Ridge/Lasso, Decision Trees, Random Forests, Support vector machines(SVM) or naive bayes.

**Supervised Classifier:** If y is categorical and discerete (e.g sapm/not, yes/no. high/medium/low), select classification models such as Logistic Regression, Decision trees, Randon Forests , svm or naive bayes.

**unsupervised learning:** if there is no target variable y, choose clustering models (e.g K-Means, DBSCAN) or dimensionality reduction techinques (e.g PCA)

# Step2.Access data complexity, linearity and size

**Linear Ralationships:** If features x exhibit a linear relationship with the target variable, simple parametric models like Linear Regression or logistic Regression are efficient and highly interpretable.

**NOn-linear/complex Relationships:** If data contains complex non-linear interactions or tabular non-linear distributions, non-parametric tree-based models like Random Forests or XGBoost perform better.

**The target variable mpg is continous. Therefore, this is a supervised regression problem, making linear regression an appropriate baseline model.

# ------------------Selecting and definig evaluation metrics-------------------

Metrics measure how well the model generalizes the unseen data. The metric depend directly on whether the task is regression or classification.

**Mean Squared Error**:

Measures the average squared differences between the actual ground truth values and predicted values.

---why use this: Squaring errors penalizes larger deviations heavily, making it sensitive to outliers.

**r2:** measures the proportion  of variance in the target variable explained by the feature inpts relative to a simple mean baseline.

---why use this: Provides a normalized score from 0.0 to 1.0(100% variance explained).

# Feature metrix x and target vector y extraction

```python
df.columns
x=df[['displacement','horsepower','acceleration']]
y=df[['mpg']]
x
y
```

**df.columns:** Return index of column names

# splitting datasets into training and testing dataset

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
y_train
```

# Model installation and training

```python
#Creates a linear Regression model
model = LinearRegression()

#Train the model on the training data
model.fit(x_train, y_train)
```

**model = LinearRegression():** Instantiates the scikit-learn LinearRegression estimator.

Sets up the ordinary least square algorithm to fit the equation.

y_cap= B_0 + B1X(disp) + B2X(horsepower) + B3X(acceleration)**model.fit(x_train, y_train):** Computes the parameter B_0(intercept) and B1,B2,B3(coefficients) by minimizing the sum of squared errors in trainig data:

# Seaborn utlities

# Executing the missing evaluation steps

```python
#Generate predictions on the held-out test set
y_pred = model.predict(x_test)

#Compute regression metrics
mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test,y_pred)

#Corrected loop using .flatten()
print("Coefficients (Betas):")
for col, coef in zip(x.columns, model.coef_.flatten()):
    print(f"  {col}: {coef:.4f}")

print("\n--- Model Performance on Test Data ---")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {np.sqrt(mse):.4f}")
print(f"R-squared Score (R2): {r2:.4f}")
```
